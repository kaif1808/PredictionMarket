# Pilot Session Review

This notebook reviews one pilot session and checks market-round data quality, price-path deviation, and volume summaries.

```python
from analysis.load import load_session
from analysis.metrics import price_path_deviation, round_volume, treatment_panel
import pandas as pd
```

```python
session_id = 1
frames = load_session(session_id)

print('markets:', len(frames.markets))
print('rounds:', len(frames.rounds))
print('trades:', len(frames.trades))
print('signals:', len(frames.signals))
```

```python
dev = price_path_deviation(frames)
vol = round_volume(frames)
panel = treatment_panel(frames)

pilot_summary = panel.groupby(['stage'], as_index=False).agg(
    mean_abs_deviation=('abs_deviation', 'mean'),
    mean_turnover=('turnover', 'mean'),
    n_rounds=('round_id', 'count')
)

pilot_summary
```

```python
data_quality_checks = {
    'has_4_markets': len(frames.markets) == 4,
    'has_20_rounds': len(frames.rounds) == 20,
    'benchmark_available_all_closed_rounds': int(frames.rounds['bayesian_benchmark'].notna().sum()) == len(frames.rounds),
    'nonnegative_trade_quantity': bool((frames.trades['quantity'] > 0).all()) if len(frames.trades) else True
}
data_quality_checks
```
